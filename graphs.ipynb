```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# ── paths ── change if needed
base = r"C:\Users\jhoni\Downloads\outputs"

def load(folder):
    for f in os.listdir(os.path.join(base, folder)):
        if f.startswith("part") and f.endswith(".csv"):
            return pd.read_csv(os.path.join(base, folder, f))

q1 = load("q1_revenue_by_category")
q2 = load("q2_event_frequency")
q3 = load("q3_sensor_alerts")

# ── Chart 1 — Revenue by Category ──
fig, ax = plt.subplots(figsize=(9, 5))
q1_sorted = q1.sort_values("total_revenue", ascending=True)
bars = ax.barh(q1_sorted["category"], q1_sorted["total_revenue"], color="#2E86AB")
ax.set_xlabel("Total Revenue (AUD)")
ax.set_title("Figure 1 — Revenue by Product Category", fontweight="bold")
for bar in bars:
    ax.text(bar.get_width() + 500, bar.get_y() + bar.get_height()/2,
            f"${bar.get_width():,.0f}", va="center", fontsize=9)
plt.tight_layout()
plt.savefig(os.path.join(base, "chart1_revenue_by_category.png"), dpi=150)
plt.close()
print("Chart 1 saved")

# ── Chart 2 — Event Type Frequency ──
fig, ax = plt.subplots(figsize=(9, 5))
q2_sorted = q2.sort_values("count", ascending=True)
ax.barh(q2_sorted["event_type"], q2_sorted["count"], color="#F18F01")
ax.set_xlabel("Number of Events")
ax.set_title("Figure 2 — Online Event Type Frequency", fontweight="bold")
for i, (val, name) in enumerate(zip(q2_sorted["count"], q2_sorted["event_type"])):
    ax.text(val + 20, i, str(val), va="center", fontsize=9)
plt.tight_layout()
plt.savefig(os.path.join(base, "chart2_event_frequency.png"), dpi=150)
plt.close()
print("Chart 2 saved")

# ── Chart 3 — Sensor Alerts by Store ──
fig, ax = plt.subplots(figsize=(9, 5))
q3_grouped = q3.groupby("store_id")["count"].sum().reset_index()
colors = ["#E84855", "#F18F01", "#2E86AB"]
ax.bar(q3_grouped["store_id"], q3_grouped["count"], color=colors)
ax.set_xlabel("Store")
ax.set_ylabel("Total Non-Normal Readings")
ax.set_title("Figure 3 — Sensor Anomalies by Store", fontweight="bold")
for i, val in enumerate(q3_grouped["count"]):
    ax.text(i, val + 0.3, str(val), ha="center", fontsize=10)
plt.tight_layout()
plt.savefig(os.path.join(base, "chart3_sensor_alerts.png"), dpi=150)
plt.close()
print("Chart 3 saved")

# ── Chart 4 — Cart Funnel ──
funnel_events = ["add_to_cart", "checkout_start", "purchase_click"]
q2_indexed = q2.set_index("event_type")
funnel_counts = [q2_indexed.loc[e, "count"] for e in funnel_events]

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(funnel_events, funnel_counts, color=["#2E86AB", "#F18F01", "#E84855"])
ax.set_ylabel("Number of Events")
ax.set_title("Figure 4 — Customer Purchase Funnel", fontweight="bold")
for bar in bars:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 10,
            str(int(bar.get_height())), ha="center", fontsize=11, fontweight="bold")
plt.tight_layout()
plt.savefig(os.path.join(base, "chart4_purchase_funnel.png"), dpi=150)
plt.close()
print("Chart 4 saved")

print("\nAll charts saved to:", base)
```
